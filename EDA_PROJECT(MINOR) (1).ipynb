```python

#univariate analysis

num_cols = ['age', 'avg_glucose_level', 'bmi',
            'hypertension', 'heart_disease', 'stroke']

for col in num_cols:
    df[col].hist(bins=30)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
#bivariate analysis
#stroke vs gender

sns.countplot(x='stroke', hue='gender', data=df)

plt.title("Stroke vs Gender")
plt.show()
```

```python
#stroke vs smoking status
sns.countplot(x='stroke', hue='smoking_status', data=df)

plt.title("Stroke vs Smoking Status")
plt.show()
```

```python
#stroke vs marital status
sns.countplot(x='stroke', hue='ever_married', data=df)

plt.title("Stroke vs Marital Status")
plt.show()
```

```python
#multivariate analysis
# Multivariate Analysis

numeric_df = df.select_dtypes(include = "number")

corr = numeric_df.corr()

sns.heatmap(corr, annot = True)

plt.show()
```

```python
df.isnull().sum()
```

```python
#missing valur treatment
#as it's a numerical value so we replace with median
#another case if there is objective value so we can replace with mode

df["bmi"] = df["bmi"].fillna(df["bmi"].median())
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["gender"] = le.fit_transform(df["gender"])
df["ever_married"] = le.fit_transform(df["ever_married"])
df["work_type"] = le.fit_transform(df["work_type"])
df["Residence_type"] = le.fit_transform(df["Residence_type"])
df["smoking_status"] = le.fit_transform(df["smoking_status"])
```

```python
df.drop('id', axis=1, inplace=True)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['age', 'avg_glucose_level', 'bmi']] = scaler.fit_transform(
    df[['age', 'avg_glucose_level', 'bmi']]
)
```
